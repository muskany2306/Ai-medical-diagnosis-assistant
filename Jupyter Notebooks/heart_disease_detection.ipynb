```python
import numpy as np
import pandas as pd
import pickle
import os

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ============================
# Load Dataset
# ============================

data = pd.read_csv("../dataset/heart_disease_data.csv")

print(data.head())

# ============================
# Features & Target
# ============================

X = data.drop(columns='target', axis=1)
Y = data['target']

# ============================
# Standardization
# ============================

scaler = StandardScaler()

X = scaler.fit_transform(X)

# ============================
# Train Test Split
# ============================

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    stratify=Y,
    random_state=2
)

# ============================
# Model Training
# ============================

model = LogisticRegression()

model.fit(X_train, Y_train)

# ============================
# Accuracy
# ============================

train_prediction = model.predict(X_train)

print("Training Accuracy :",
      accuracy_score(Y_train, train_prediction))

test_prediction = model.predict(X_test)

print("Testing Accuracy :",
      accuracy_score(Y_test, test_prediction))

# ============================
# Save Model & Scaler
# ============================

os.makedirs("../Models", exist_ok=True)

pickle.dump(model,
            open("../Models/heart_disease_model.sav", "wb"))

pickle.dump(scaler,
            open("../Models/scaler.sav", "wb"))

print("Model Saved Successfully")

# ============================
# Load Model
# ============================

loaded_model = pickle.load(
    open("../Models/heart_disease_model.sav", "rb")
)

loaded_scaler = pickle.load(
    open("../Models/scaler.sav", "rb")
)

print("Model Loaded Successfully")

# ============================
# Prediction
# ============================

input_data = (
    57,
    0,
    0,
    120,
    354,
    0,
    1,
    163,
    1,
    0.6,
    2,
    0,
    2
)

input_array = np.asarray(input_data).reshape(1, -1)

input_scaled = loaded_scaler.transform(input_array)

prediction = loaded_model.predict(input_scaled)

if prediction[0] == 0:
    print("The Person does NOT have Heart Disease")
else:
    print("The Person HAS Heart Disease")
```
